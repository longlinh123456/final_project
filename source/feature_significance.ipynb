```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

df = pd.read_csv("../encoded_dataset.csv")

# >= 4 is 1 (procrastinate), < 4 is 0 (does not procrastinate)
y = (df["delay_until_deadline"] >= 4).astype(int)

# Drop target and insignificant columns to create feature set
features_to_drop = [
    "delay_until_deadline",
    "timestamp",
    "referrer",
    "data_quality_check",
]
X_raw = df.drop(columns=features_to_drop)

# Impute missing values (median)
# Should not matter since there is only 1 missing value in 1 column
imputer = SimpleImputer(strategy="median")
X_imputed = pd.DataFrame(imputer.fit_transform(X_raw), columns=X_raw.columns)

# Scaling for Logistic Regression
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X_imputed), columns=X_raw.columns)

# Logistic Regression Coefficients
lr = LogisticRegression(random_state=67)
lr.fit(X_scaled, y)
lr_importances = pd.Series(np.abs(lr.coef_[0]), index=X_scaled.columns).sort_values(
    ascending=False
)
lr_signs = pd.Series(lr.coef_[0], index=X_scaled.columns)  # To see direction

print("\nTop Features (Logistic Regression Absolute Coefficients):")
for feat in lr_importances.index:
    sign = "Positive" if lr_signs[feat] > 0 else "Negative"
    print(f"{feat:35} {lr_importances[feat]:.4f} ({sign} correlation)")
```
